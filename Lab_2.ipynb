```python
#depth first search

def dfs(start, goal):
    stack = [(start, [start])]
    visited = set()

    while stack:
        state, path = stack.pop()

        if state == goal:
            return path

        if state in visited:
            continue

        visited.add(state)
        zero = state.index(0)
        row, col = divmod(zero, 3)

        moves = [
            (row - 1, col),
            (row + 1, col),
            (row, col - 1),
            (row, col + 1)
        ]

        for r, c in moves:
            if 0 <= r < 3 and 0 <= c < 3:
                new_pos = r * 3 + c
                new_state = list(state)
                new_state[zero], new_state[new_pos] = (
                    new_state[new_pos], new_state[zero]
                )
                new_state = tuple(new_state)

                if new_state not in visited:
                    stack.append((new_state, path + [new_state]))

    return None


start = (1, 2, 3,
         4, 5, 6,
         0, 7, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

result = dfs(start, goal)

if result:
    print("Solution found!")
    print("Number of moves:", len(result) - 1)

    for state in result:
        for i in range(0, 9, 3):
            print(state[i:i + 3])
        print()
else:
    print("No solution found.")
```

```python
#iterative deepning search

def dls(state, goal, limit, path, visited):
    if state == goal:
        return path

    if limit == 0:
        return None

    zero = state.index(0)
    row, col = divmod(zero, 3)

    moves = [
        (row - 1, col),
        (row + 1, col),
        (row, col - 1),
        (row, col + 1)
    ]

    for r, c in moves:
        if 0 <= r < 3 and 0 <= c < 3:
            new_pos = r * 3 + c
            new_state = list(state)
            new_state[zero], new_state[new_pos] = (
                new_state[new_pos], new_state[zero]
            )
            new_state = tuple(new_state)

            if new_state not in visited:
                visited.add(new_state)

                result = dls(
                    new_state, goal, limit - 1,
                    path + [new_state], visited
                )

                if result:
                    return result

    return None


def ids(start, goal):
    for depth in range(100):
        result = dls(start, goal, depth, [start], {start})

        if result:
            return result

    return None


start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

result = ids(start, goal)

if result:
    print("Solution found!")
    print("Number of moves:", len(result) - 1)

    for state in result:
        for i in range(0, 9, 3):
            print(state[i:i + 3])
        print()
else:
    print("No solution found within the depth limit.")
```
