```python
import random
import math

def cost(s):
    return sum(s[i] == s[j] or
               abs(s[i]-s[j]) == abs(i-j)
               for i in range(8) for j in range(i+1, 8))

s = [random.randrange(8) for _ in range(8)]
T = 100

while T > 0.01 and cost(s) != 0:
    n = s[:]
    c = random.randrange(8)
    n[c] = random.randrange(8)

    d = cost(s) - cost(n)

    if d > 0 or random.random() < math.exp(d/T):
        s = n

    T *= 0.99

print("Board:", s)
print("Cost:", cost(s))
```
