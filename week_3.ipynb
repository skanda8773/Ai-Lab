```python
#A* using misplaced tiles

import heapq

goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)

def h(s):
    return sum(s[i] != goal[i] and s[i] != 0 for i in range(9))

def astar(start):
    pq = [(h(start), 0, start, [])]
    visited = set()

    while pq:
        f, g, s, path = heapq.heappop(pq)

        if s == goal:
            return path + [s]

        if s in visited:
            continue
        visited.add(s)

        z = s.index(0)
        r, c = divmod(z, 3)

        for dr, dc in [(1,0), (-1,0), (0,1), (0,-1)]:
            nr, nc = r + dr, c + dc
            if 0 <= nr < 3 and 0 <= nc < 3:
                nz = nr * 3 + nc
                t = list(s)
                t[z], t[nz] = t[nz], t[z]
                t = tuple(t)

                if t not in visited:
                    heapq.heappush(
                        pq, (g + 1 + h(t), g + 1, t, path + [s])
                    )
    return None

start = tuple(map(int, input().split()))
ans = astar(start)

if ans:
    for s in ans:
        print(s[:3])
        print(s[3:6])
        print(s[6:])
        print()
    print("Steps:", len(ans) - 1)
else:
    print("No solution")
```

```python
#A* using manhattan distance

import heapq

goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)

def h(s):
    d = 0
    for i, x in enumerate(s):
        if x:
            j = x - 1
            r1, c1 = divmod(i, 3)
            r2, c2 = divmod(j, 3)
            d += abs(r1-r2) + abs(c1-c2)
    return d

def astar(start):
    pq = [(h(start), 0, start, [])]
    visited = set()

    while pq:
        f, g, s, path = heapq.heappop(pq)

        if s == goal:
            return path + [s]

        if s in visited:
            continue
        visited.add(s)

        z = s.index(0)
        r, c = divmod(z, 3)

        for dr, dc in [(1,0), (-1,0), (0,1), (0,-1)]:
            nr, nc = r + dr, c + dc
            if 0 <= nr < 3 and 0 <= nc < 3:
                nz = nr * 3 + nc
                t = list(s)
                t[z], t[nz] = t[nz], t[z]
                t = tuple(t)

                if t not in visited:
                    heapq.heappush(
                        pq, (g + 1 + h(t), g + 1, t, path + [s])
                    )
    return None

start = tuple(map(int, input().split()))
ans = astar(start)

if ans:
    for s in ans:
        print(s[:3])
        print(s[3:6])
        print(s[6:])
        print()
    print("Steps:", len(ans) - 1)
else:
    print("No solution")
```
