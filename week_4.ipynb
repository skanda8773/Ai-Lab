```python
from itertools import combinations
def cost(s):
  return sum(s[i]==s[j] or abs(s[i]-s[j])== abs(i-j)
  for i in range (4) for j in range(i+1,4))
s=[3,1,2,0]
print("initial:", s, "cost:", cost(s))
while cost(s)>0:
  best=s[:]
  for i,j in  combinations(range(4),2):
    n=s[:]
    n[i],n[j]=n[j],n[i]
    if cost(n)<cost(best):
      best=n
  if cost(best)>=cost(s):
    print("local minimum reached")
    break
  s=best
  print("next:",s,"cost:",cost(s))
  print("final state:",s)
  print("final cost:",cost(s))
```
